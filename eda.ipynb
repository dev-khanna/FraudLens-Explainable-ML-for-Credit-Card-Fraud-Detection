```python
# Why duplicates matter: 
# the primary risk is train/test leakage. If a row appears 
# twice and one copy ends up in your training set while the other ends up in your 
# test set, the model isn't being tested on unseen data anymore. It's literally 
# being asked to recognize a row it already memorized during training.

# That makes your test-set precision/recall look better than the model would 
# actually perform on truly new transactions. It's a smaller-scale version of 
# the same leakage problem the temporal split was designed to prevent.

# On top of that, duplicates quietly change your class balance numbers (if some 
# of the 492 fraud rows are duplicates, your "true" number of distinct fraud 
# cases is lower than 492) and give repeated rows extra influence on anything 
# averaged, like the mean or the model's loss.
```

```python
fraud_times = df[df['Class'] == 1]['Time'].sort_values()
gaps = fraud_times.diff()

print(gaps.describe())
print('within 60 sec of previous fraud:', (gaps < 60).sum())
print('within 10 sec of previous fraud:', (gaps < 10).sum())
print('fraction within 10 sec:', (gaps < 10).sum() / len(fraud_times))

# group consecutive frauds into bursts using a 10-second gap threshold
burst_id = (gaps >= 10).cumsum()
burst_sizes = burst_id.value_counts()
burst_sizes.describe()
burst_sizes.sort_values(ascending=False).head(10)
```

```python
print(burst_sizes.describe())
print('total distinct bursts:', burst_id.nunique())
print('bursts with size >= 2:', (burst_sizes >= 2).sum())
print(burst_sizes.value_counts().sort_index())
```

```python
df_fraud = df[df['Class'] == 1].sort_values('Time').copy()
df_fraud['gap'] = df_fraud['Time'].diff()
df_fraud['burst_id'] = (df_fraud['gap'] >= 10).cumsum()
burst_sizes_full = df_fraud.groupby('burst_id').size()
df_fraud['burst_size'] = df_fraud['burst_id'].map(burst_sizes_full)
df_fraud['is_clustered'] = df_fraud['burst_size'] > 1

pd.crosstab(df_fraud['hour_of_day'], df_fraud['is_clustered'])
```

```python
if 'day' not in df_fraud.columns:
    df_fraud['day'] = (df_fraud['Time'] // 86400).astype(int)

# collapse each burst down to a single event
burst_events = df_fraud.drop_duplicates(subset='burst_id', keep='first')

event_counts = burst_events.groupby(['day', 'hour_of_day']).size().reset_index(name='event_count')
event_counts = event_counts.merge(
    day_hour['count'].reset_index(), on=['day', 'hour_of_day']
)
event_counts['event_rate'] = event_counts['event_count'] / event_counts['count']

event_counts.pivot(index='hour_of_day', columns='day', values='event_rate')
```

```python
early_hours = [1, 2, 3, 4]
window_check = event_counts[event_counts['hour_of_day'].isin(early_hours)].copy()
window_check.groupby('day')[['event_count', 'count']].sum().assign(
    window_rate=lambda d: d['event_count'] / d['count']
)
```

```python
window_events = window_check['event_count'].sum()
window_txns = window_check['count'].sum()
window_rate = window_events / window_txns

total_events = burst_id.nunique()
total_txns = len(df)

non_window_events = total_events - window_events
non_window_txns = total_txns - window_txns
non_window_rate = non_window_events / non_window_txns

print(f'window: {window_events}/{window_txns} = {window_rate:.5f}')
print(f'outside window: {non_window_events}/{non_window_txns} = {non_window_rate:.5f}')
print(f'ratio: {window_rate / non_window_rate:.2f}x')

# two-proportion z-test, window vs outside
import numpy as np
p_pool = total_events / total_txns
se = np.sqrt(p_pool * (1 - p_pool) * (1/window_txns + 1/non_window_txns))
z = (window_rate - non_window_rate) / se
print(f'z = {z:.2f}')
```

```python
# early-morning window (hours 1-4) has an event rate ~4.5x the rest of the dataset
# (z~13, not a small-sample fluke). the two days agree with each other on this -
# the 24% gap between day0 and day1 rates is noise (z~0.9, not significant).
# NOT using single-hour granularity (hour 2 specifically) - that didn't survive
# burst dedup and shouldn't be trusted as a feature boundary.
```

```python
# Target: 283,253 legit / 473 fraud (0.1667%) after removing 1,081 duplicates
# Amount: heavy overlap between classes, not a strong standalone predictor
# Features: V14/V17/V12/V10/V16/V11/V3/V4 show the strongest class separation
#           (ranked by standardized median difference, not raw correlation)
# Time: early-morning window (hours 1-4) has ~4.5x baseline event rate,
#       confirmed via burst-deduplication + cross-day comparison + z-test (z~13)
#       -> engineer as is_early_morning, NOT a single-hour dummy
# Bursts: 415 distinct fraud events, 375 singletons, max burst size 5 ->
#         bounded leakage risk, not structural
```

```python
# 1) does any burst straddle the day boundary? if a temporal split cuts at day0/day1,
#    this tells us whether that cut would leak a burst across it - same problem
#    burst analysis already flagged for random splits.

df_fraud['day'] = (df_fraud['Time'] // 86400).astype(int)
straddling = df_fraud.groupby('burst_id')['day'].nunique()
(straddling > 1).sum()
```

```python

# 2) do the top separating features actually drift between the two days,
#    or is the fraud "profile" stable? (the real justification a temporal split needs)
from scipy.stats import ks_2samp
for col in ['V14', 'V17', 'V12', 'V10']:
    d0 = df_fraud[df_fraud['day'] == 0][col]
    d1 = df_fraud[df_fraud['day'] == 1][col]
    stat, p = ks_2samp(d0, d1)
    print(col, 'KS stat:', round(stat, 3), 'p-value:', round(p, 3))
```

```python
df['burst_id'] = pd.NA
df.loc[df_fraud.index, 'burst_id'] = df_fraud['burst_id']

# legit rows: each gets its own group so none are tied together
legit_mask = df['burst_id'].isna()
df.loc[legit_mask, 'burst_id'] = 'legit_' + df.index[legit_mask].astype(str)

dev = df[df['day'] == 0].copy()
test = df[df['day'] == 1].copy()

print(dev['burst_id'].nunique(), 'groups in dev set')
```

```python
## Modeling

from sklearn.model_selection import StratifiedGroupKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)


# 1. Prepare development and final test sets


dev['burst_id'] = dev['burst_id'].astype(str)
test['burst_id'] = test['burst_id'].astype(str)

v_cols = [f'V{i}' for i in range(1, 29)]

feature_cols = v_cols + ['Amount', 'hour_of_day']

X_dev = dev[feature_cols]
y_dev = dev['Class']
groups_dev = dev['burst_id']

X_test = test[feature_cols]
y_test = test['Class']



# 2. Preprocessing

# V1-V28 are already PCA-derived features.
# Amount and hour_of_day are scaled using RobustScaler.
#
# Putting preprocessing inside the Pipeline ensures that
# scaling is fitted only on the training fold.

preprocess = ColumnTransformer(
    transformers=[
        ('scale', RobustScaler(), ['Amount', 'hour_of_day'])
    ],
    remainder='passthrough'
)



# 3. Logistic Regression baseline


logistic_model = Pipeline(
    steps=[
        ('prep', preprocess),
        (
            'clf',
            LogisticRegression(
                class_weight='balanced',
                max_iter=1000,
                random_state=42
            )
        )
    ]
)



# 4. Group-aware cross-validation on development data


cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_validate(
    logistic_model,
    X_dev,
    y_dev,
    groups=groups_dev,
    cv=cv,
    scoring={
        'average_precision': 'average_precision',
        'roc_auc': 'roc_auc',
        'precision': 'precision',
        'recall': 'recall',
        'f1': 'f1'
    },
    n_jobs=-1
)


# 5. Display cross-validation results

print("Logistic Regression - 5-Fold Group-Aware CV")
print("-" * 50)

for metric in [
    'test_average_precision',
    'test_roc_auc',
    'test_precision',
    'test_recall',
    'test_f1'
]:
    mean_score = cv_scores[metric].mean()
    std_score = cv_scores[metric].std()

    print(
        f"{metric.replace('test_', '').upper():20s}: "
        f"{mean_score:.4f} ± {std_score:.4f}"
    )
```

```python
# 6. Train Logistic Regression on the full development set

logistic_model.fit(X_dev, y_dev)


# 7. Evaluate on untouched Day-1 test set

test_probabilities = logistic_model.predict_proba(X_test)[:, 1]

# Default threshold = 0.50
test_predictions = (test_probabilities >= 0.50).astype(int)


# 8. Test-set metrics

average_precision = average_precision_score(
    y_test,
    test_probabilities
)

roc_auc = roc_auc_score(
    y_test,
    test_probabilities
)

precision = precision_score(
    y_test,
    test_predictions,
    zero_division=0
)

recall = recall_score(
    y_test,
    test_predictions,
    zero_division=0
)

f1 = f1_score(
    y_test,
    test_predictions,
    zero_division=0
)

cm = confusion_matrix(
    y_test,
    test_predictions
)


print("\nLogistic Regression - Final Day-1 Test")
print("-" * 50)
print(f"Average Precision (PR-AUC): {average_precision:.4f}")
print(f"ROC-AUC:                   {roc_auc:.4f}")
print(f"Precision:                 {precision:.4f}")
print(f"Recall:                    {recall:.4f}")
print(f"F1 Score:                  {f1:.4f}")

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_predictions,
        digits=4,
        zero_division=0
    )
)
```

```python

# XGBoost Model


from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedGroupKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler
```

```python

# 1. Calculate scale_pos_weight

# This accounts for the severe class imbalance.

negative_count = (y_dev == 0).sum()
positive_count = (y_dev == 1).sum()

scale_pos_weight = negative_count / positive_count

print(f"Negative samples: {negative_count}")
print(f"Positive samples: {positive_count}")
print(f"scale_pos_weight: {scale_pos_weight:.2f}")
```

```python

# 2. Build XGBoost pipeline


xgb_model = Pipeline(
    steps=[
        ('prep', preprocess),
        (
            'clf',
            XGBClassifier(
                n_estimators=300,
                max_depth=4,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                scale_pos_weight=scale_pos_weight,
                objective='binary:logistic',
                eval_metric='logloss',
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python

# 3. Group-aware cross-validation


xgb_cv_scores = cross_validate(
    xgb_model,
    X_dev,
    y_dev,
    groups=groups_dev,
    cv=cv,
    scoring={
        'average_precision': 'average_precision',
        'roc_auc': 'roc_auc',
        'precision': 'precision',
        'recall': 'recall',
        'f1': 'f1'
    },
    n_jobs=-1
)
```

```python

# 4. Display XGBoost CV results


print("XGBoost - 5-Fold Group-Aware CV")
print("-" * 50)

for metric in [
    'test_average_precision',
    'test_roc_auc',
    'test_precision',
    'test_recall',
    'test_f1'
]:
    mean_score = xgb_cv_scores[metric].mean()
    std_score = xgb_cv_scores[metric].std()

    print(
        f"{metric.replace('test_', '').upper():20s}: "
        f"{mean_score:.4f} ± {std_score:.4f}"
    )
```

```python

# 5. Train XGBoost on the full development set


xgb_model.fit(X_dev, y_dev)



# 6. Predict fraud probabilities on Day-1 test set


xgb_test_probabilities = xgb_model.predict_proba(X_test)[:, 1]

# Default threshold = 0.50
xgb_test_predictions = (
    xgb_test_probabilities >= 0.50
).astype(int)



# 7. Evaluate XGBoost on Day-1 test set


xgb_average_precision = average_precision_score(
    y_test,
    xgb_test_probabilities
)

xgb_roc_auc = roc_auc_score(
    y_test,
    xgb_test_probabilities
)

xgb_precision = precision_score(
    y_test,
    xgb_test_predictions,
    zero_division=0
)

xgb_recall = recall_score(
    y_test,
    xgb_test_predictions,
    zero_division=0
)

xgb_f1 = f1_score(
    y_test,
    xgb_test_predictions,
    zero_division=0
)

xgb_cm = confusion_matrix(
    y_test,
    xgb_test_predictions
)


print("\nXGBoost - Final Day-1 Test")
print("-" * 50)
print(f"Average Precision (PR-AUC): {xgb_average_precision:.4f}")
print(f"ROC-AUC:                   {xgb_roc_auc:.4f}")
print(f"Precision:                 {xgb_precision:.4f}")
print(f"Recall:                    {xgb_recall:.4f}")
print(f"F1 Score:                  {xgb_f1:.4f}")

print("\nConfusion Matrix:")
print(xgb_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        xgb_test_predictions,
        digits=4,
        zero_division=0
    )
)
```

```python

# Threshold Tuning


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import cross_val_predict



# 1. Get out-of-fold fraud probabilities

# These predictions come from models that did NOT train on
# the corresponding validation rows.

dev_oof_probabilities = cross_val_predict(
    xgb_model,
    X_dev,
    y_dev,
    groups=groups_dev,
    cv=cv,
    method='predict_proba',
    n_jobs=-1
)[:, 1]



# 2. Evaluate different probability thresholds


thresholds = np.arange(0.05, 0.96, 0.05)

threshold_results = []

for threshold in thresholds:

    predictions = (
        dev_oof_probabilities >= threshold
    ).astype(int)

    precision = precision_score(
        y_dev,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_dev,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_dev,
        predictions,
        zero_division=0
    )

    threshold_results.append({
        'threshold': threshold,
        'precision': precision,
        'recall': recall,
        'f1': f1
    })


threshold_results = pd.DataFrame(threshold_results)

print(threshold_results)
```

```python

# 3. Plot precision-recall trade-off


plt.figure(figsize=(10, 6))

plt.plot(
    threshold_results['threshold'],
    threshold_results['precision'],
    marker='o',
    label='Precision'
)

plt.plot(
    threshold_results['threshold'],
    threshold_results['recall'],
    marker='o',
    label='Recall'
)

plt.plot(
    threshold_results['threshold'],
    threshold_results['f1'],
    marker='o',
    label='F1'

)

plt.xlabel('Decision Threshold')
plt.ylabel('Score')
plt.title('XGBoost Threshold Tuning')
plt.legend()
plt.grid(True)
plt.show()
```

```python

# 4. Find the threshold with the highest F1


best_threshold = threshold_results.loc[
    threshold_results['f1'].idxmax(),
    'threshold'
]

best_f1 = threshold_results.loc[
    threshold_results['f1'].idxmax(),
    'f1'
]

print(f"Best threshold: {best_threshold:.2f}")
print(f"Best CV F1:     {best_f1:.4f}")
```

```python

# Evaluate the tuned threshold on the untouched Day-1 test


best_threshold = 0.80

# Convert predicted probabilities into fraud/legitimate labels
xgb_test_predictions_tuned = (
    xgb_test_probabilities >= best_threshold
).astype(int)



# Calculate metrics


tuned_precision = precision_score(
    y_test,
    xgb_test_predictions_tuned,
    zero_division=0
)

tuned_recall = recall_score(
    y_test,
    xgb_test_predictions_tuned,
    zero_division=0
)

tuned_f1 = f1_score(
    y_test,
    xgb_test_predictions_tuned,
    zero_division=0
)

tuned_cm = confusion_matrix(
    y_test,
    xgb_test_predictions_tuned
)



# Display results


print("XGBoost - Tuned Threshold (0.80) - Day-1 Test")
print("-" * 55)

print(f"Threshold: {best_threshold:.2f}")
print(f"Precision: {tuned_precision:.4f}")
print(f"Recall:    {tuned_recall:.4f}")
print(f"F1 Score:  {tuned_f1:.4f}")

print("\nConfusion Matrix:")
print(tuned_cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        xgb_test_predictions_tuned,
        digits=4,
        zero_division=0
    )
)
```

```python

# XGBoost Hyperparameter Tuning


from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform, loguniform



# 1. Define the parameter search space


param_distributions = {

    # Number of trees
    'clf__n_estimators': randint(200, 600),

    # Maximum depth of each tree
    'clf__max_depth': randint(3, 8),

    # Learning rate
    'clf__learning_rate': loguniform(0.01, 0.15),

    # Minimum loss reduction required for another split
    'clf__gamma': uniform(0, 0.5),

    # Minimum number of samples/weight in a child node
    'clf__min_child_weight': randint(1, 10),

    # Fraction of training rows used for each boosting round
    'clf__subsample': uniform(0.7, 0.3),

    # Fraction of features used for each tree
    'clf__colsample_bytree': uniform(0.7, 0.3),

    # L1 regularization
    'clf__reg_alpha': loguniform(1e-4, 1.0),

    # L2 regularization
    'clf__reg_lambda': loguniform(0.1, 10.0),

    # Class imbalance adjustment
    'clf__scale_pos_weight': [
        scale_pos_weight * 0.75,
        scale_pos_weight,
        scale_pos_weight * 1.25
    ]
}



# 2. Randomized hyperparameter search


xgb_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_distributions,
    n_iter=20,
    scoring='average_precision',
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)



# 3. Run search ONLY on the development data


xgb_search.fit(
    X_dev,
    y_dev,
    groups=groups_dev
)



# 4. Display best result


print("\nBest Cross-Validation PR-AUC:")
print(f"{xgb_search.best_score_:.4f}")

print("\nBest Hyperparameters:")
for parameter, value in xgb_search.best_params_.items():
    print(f"{parameter}: {value}")
```
